### About the dataset

You will be working with the King County House Sales dataset. This dataset contains historical sales data for houses in King County, Washington, USA, including those in Seattle. **Dataset Link**: https://www.kaggle.com/datasets/harlfoxem/housesalesprediction

**Column Descriptions**:

- **id**: A unique identifier for each house sale record.
- **date**: The date when the house was sold.
- **price**: The sale price of the house (target variable for prediction).
- **bedrooms**: Number of bedrooms in the house.
- **bathrooms**: Number of bathrooms in the house.
- **sqft_living**: Square footage of the interior living space.
- **sqft_lot**: Square footage of the land lot.
- **floors**: Number of floors (levels) in the house.
- **waterfront**: A binary indicator (1 if the house has a waterfront view, 0 otherwise).
- **view**: Not sure about this one. The number of views by the potential buyers or an ordinal variable indicating the quality of the view (e.g., 0 for no view, higher values for better views).
- **condition**: An ordinal variable indicating the overall condition of the house (e.g., 1-5, with 5 being excellent).
- **grade**: An overall grade given to the housing unit, based on the King County grading system (e.g., higher values indicate higher quality construction and design). 
- **sqft_above**: Square footage of the house's living space above ground.
- **sqft_basement**: Square footage of the basement area.
- **yr_built**: The year the house was built.
- **yr_renovated**: The year the house was last renovated. (A value of 0 typically means it has not been renovated).
- **zipcode**: The 5-digit ZIP code of the property.
- **lat**: Latitude coordinate of the house's location.
- **long**: Longitude coordinate of the house's location.
- **sqft_living15**: The average square footage of interior living space for the 15 nearest neighbours in 2015.
- **sqft_lot15**: The average square footage of land lots for the 15 nearest neighbours in 2015.
