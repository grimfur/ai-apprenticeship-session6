Two new features were created from the existing dataset to improve the potentially improve the usefulness of the data for future modelling tasks.

1. Genre Count

The genre_and_votes column contains multiple genres stored in a comma-separated text format. To better represent this information, a new column called genre_count was created to capture how many genres are associated with each book.

books['genre_count'] = books['genre_and_votes'].dropna().str.split(',').str.len()

This feature indicates the number of genres assigned to a book. Books associated with multiple genres may appeal to a wider audience and could influence recommendation patterns.

2. Positive Rating Ratio
A new feature called positive_rating_ratio was created using the star rating distribution columns.

books['positive_rating_ratio'] = (
    books['four_star_ratings'] + books['five_star_ratings']
) / books['rating_count']

This feature represents the proportion of highly positive ratings relative to the total number of ratings.

Compared with raw counts, this ratio provides a clearer signal of how positively readers perceive a book. It helps distinguish between books that are widely rated and books that are consistently rated highly.

Why These Features Are Useful

These engineered features provide more meaningful information than the original raw fields.

Feature	Benefit
genre_count	captures how broadly a book spans different genres
positive_rating_ratio	measures reader sentiment more clearly

These help convert raw data into structured signals that are more suitable for machine learning models, particularly recommendation systems.
